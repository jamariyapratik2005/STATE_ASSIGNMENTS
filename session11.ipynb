# Lists and Matrices in Real-World Applications

This notebook uses Python lists and matrices for playlist, food ratings, order totals, and product revenue calculations.

```python
# 1. Total duration of a playlist
playlist_durations = [210, 180, 245, 200, 175]
total_duration = sum(playlist_durations)

print(f"Song durations: {playlist_durations} seconds")
print(f"Total playlist duration: {total_duration} seconds")
```

```python
# 2. Ratings given by 3 friends to 3 Swiggy food items
ratings_matrix = [
    [5, 4, 3],
    [4, 5, 4],
    [3, 4, 5]
]

print("Ratings matrix:")
for row in ratings_matrix:
    print(row)

print("Ratings each friend gave to the second food item:")
for friend_number, row in enumerate(ratings_matrix, start=1):
    print(f"Friend {friend_number}: {row[1]}")
```

```python
# 3. Total orders for each dish across 3 days
# Rows represent days; columns represent dishes.
orders_matrix = [
    [12, 8, 10, 6],
    [15, 9, 7, 11],
    [10, 13, 12, 8]
]

total_orders_per_dish = [
    sum(day_orders[dish_index] for day_orders in orders_matrix)
    for dish_index in range(4)
]

print("Orders matrix:")
for row in orders_matrix:
    print(row)
print(f"Total orders for each dish: {total_orders_per_dish}")
```

```python
# 4. Total revenue for 5 Flipkart products
products_matrix = [
    [499.00, 20],
    [1299.00, 15],
    [799.00, 25],
    [2499.00, 8],
    [349.00, 30]
]


def calculate_product_revenue(products):
    """Return price multiplied by quantity sold for each product."""
    return [price * quantity for price, quantity in products]


revenues = calculate_product_revenue(products_matrix)
print("Product matrix (price, quantity sold):")
for product_number, product in enumerate(products_matrix, start=1):
    print(f"Product {product_number}: {product}")
print(f"Total revenue for each product: {revenues}")
```
