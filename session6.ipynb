## 2. Zomato Home Page Experiment

The difference between the sample means is ₹30. Using a two-sample t-test with equal sample sizes and standard deviations:

- Group A mean = ₹350, Group B mean = ₹320
- Sample sizes: $n_A = n_B = 30$
- Standard deviations: $s_A = s_B = ₹50$
- Test statistic: $t \approx 2.324$
- Two-sided p-value: $p \approx 0.024$

Since the p-value (0.024) is less than 0.05, reject H0 at the 5% significance level. There is statistically significant evidence that the average order values of the two groups are different.


```python
import math
from scipy import stats

mean_a = 350
mean_b = 320
std_a = 50
std_b = 50
n_a = 30
n_b = 30

# Calculate t-statistic
t_statistic = (mean_a - mean_b) / math.sqrt(
    (std_a**2 / n_a) + (std_b**2 / n_b)
)

# Degrees of freedom
df = n_a + n_b - 2

# Calculate two-sided p-value
p_value = 2 * (1 - stats.t.cdf(abs(t_statistic), df))

print(f"t-statistic: {t_statistic:.3f}")
print(f"p-value: {p_value:.3f}")

if p_value < 0.05:
    print("Decision: Reject H0")
else:
    print("Decision: Fail to reject H0")
```

## 3. WhatsApp Dark Mode A/B Test

- **Control group:** A randomly selected group continues using the existing WhatsApp light mode.
- **Variant group:** A randomly selected group is shown the new dark mode.
- **Metric:** Measure the average daily session time or the percentage of users who return the next day over a fixed test period.
- **Null hypothesis (H0):** The new dark mode does not change the chosen metric compared with light mode.
- **Alternative hypothesis (H1):** The new dark mode increases the chosen metric compared with light mode.

Users should be randomly assigned to the two groups, and the groups should be compared over the same time period.

## 4. Flipkart Wishlist Button

The difference is 520 clicks versus 480 clicks, or 52% versus 48%, a 4 percentage-point difference. This difference might be due to chance, so it cannot be called significant from the counts alone. The p-value measures how likely a difference this large would be if the button designs truly had the same click rate.
