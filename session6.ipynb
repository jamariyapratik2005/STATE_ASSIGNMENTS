# Hypothesis Testing and A/B Testing

## 1. Instagram vs YouTube Usage

- **Null hypothesis (H0):** People do not spend more time on Instagram than on YouTube daily. In symbols, $\mu_{Instagram} \leq \mu_{YouTube}$.
- **Alternative hypothesis (H1):** People spend more time on Instagram than on YouTube daily. In symbols, $\mu_{Instagram} > \mu_{YouTube}$.

## 2. Zomato Home Page Experiment

The difference between the sample means is ₹30. Using a two-sample t-test with equal sample sizes and standard deviations:

- Group A mean = ₹350, Group B mean = ₹320
- Sample sizes: $n_A = n_B = 30$
- Standard deviations: $s_A = s_B = ₹50$
- Test statistic: $t \approx 2.324$
- Two-sided p-value: $p \approx 0.024$

Since $p < 0.05$, reject the null hypothesis at the 5% significance level. There is statistically significant evidence that the new Zomato home page changed the average order value. The code cell below calculates the p-value.

## 3. WhatsApp Dark Mode A/B Test

- **Control group:** A randomly selected group continues using the existing WhatsApp light mode.
- **Variant group:** A randomly selected group is shown the new dark mode.
- **Metric:** Measure the average daily session time or the percentage of users who return the next day over a fixed test period.
- **Null hypothesis (H0):** The new dark mode does not change the chosen metric compared with light mode.
- **Alternative hypothesis (H1):** The new dark mode increases the chosen metric compared with light mode.

Users should be randomly assigned to the two groups, and the groups should be compared over the same time period.

## 4. Flipkart Wishlist Button

The difference is 520 clicks versus 480 clicks, or 52% versus 48%, a 4 percentage-point difference. This difference might be due to chance, so it cannot be called significant from the counts alone. The p-value measures how likely a difference this large would be if the button designs truly had the same click rate. If the p-value is below 0.05, reject the null hypothesis and call the result statistically significant; otherwise, do not reject the null hypothesis.

import scipy.stats as stats

mean_a, mean_b = 350, 320
std_a, std_b = 50, 50
n_a, n_b = 30, 30

# Reconstruct summary statistics as repeated observations for scipy's t-test.
group_a = [mean_a - std_a] * 15 + [mean_a + std_a] * 15
group_b = [mean_b - std_b] * 15 + [mean_b + std_b] * 15

t_statistic, p_value = stats.ttest_ind(group_a, group_b, equal_var=True)

print(f"t-statistic: {t_statistic:.3f}")
print(f"two-sided p-value: {p_value:.3f}")
print("Decision: Reject H0" if p_value < 0.05 else "Decision: Do not reject H0")

```python
import scipy.stats as stats
import math

mean_a, mean_b = 350, 320
std_a, std_b = 50, 50
n_a, n_b = 30, 30

# Build illustrative data with the requested means and sample standard deviations.
spread_a = std_a * math.sqrt((n_a - 1) / n_a)
spread_b = std_b * math.sqrt((n_b - 1) / n_b)
group_a = [mean_a - spread_a] * (n_a // 2) + [mean_a + spread_a] * (n_a // 2)
group_b = [mean_b - spread_b] * (n_b // 2) + [mean_b + spread_b] * (n_b // 2)

t_statistic, p_value = stats.ttest_ind(group_a, group_b, equal_var=True)

print(f"t-statistic: {t_statistic:.3f}")
print(f"two-sided p-value: {p_value:.3f}")
print("Decision: Reject H0" if p_value < 0.05 else "Decision: Do not reject H0")
```
