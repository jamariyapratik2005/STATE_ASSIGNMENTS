# Probability in Python

This notebook demonstrates five probability problems using functions, simulation, lists, and sets.

```python
# 1. Calculate the probability of getting Heads on a fair coin toss

def calculate_probability(event_count, total_outcomes):
    if total_outcomes <= 0:
        raise ValueError("total_outcomes must be greater than zero")
    return event_count / total_outcomes


heads_probability = calculate_probability(1, 2)
print(f"Probability of Heads: {heads_probability:.2f}")
```

```python
# 2. Simulate rolling a six-sided die 100 times
import random


def calculate_probability(event_count, total_outcomes):
    if total_outcomes <= 0:
        raise ValueError("total_outcomes must be greater than zero")
    return event_count / total_outcomes


random.seed(42)
dice_rolls = [random.randint(1, 6) for _ in range(100)]
sixes_count = dice_rolls.count(6)
experimental_probability = calculate_probability(sixes_count, len(dice_rolls))

print(f"Number of 6s: {sixes_count}")
print(f"Experimental probability of rolling a 6: {experimental_probability:.2f}")
```

```python
# 3. Probability of randomly selecting a pop song

def calculate_probability(event_count, total_outcomes):
    if total_outcomes <= 0:
        raise ValueError("total_outcomes must be greater than zero")
    return event_count / total_outcomes


recent_genres = [
    "pop", "rock", "pop", "hiphop", "jazz",
    "pop", "rock", "pop", "classical", "pop",
    "hiphop", "rock", "pop", "jazz", "pop",
    "rock", "pop", "country", "pop", "rock"
]
pop_count = recent_genres.count("pop")
pop_probability = calculate_probability(pop_count, len(recent_genres))

print(f"Pop songs: {pop_count} out of {len(recent_genres)}")
print(f"Probability of selecting a pop song: {pop_probability:.2f}")
```

```python
# 4. Probability that a user ordered both food and dessert

def probability_of_both(food_users, dessert_users):
    food_users = set(food_users)
    dessert_users = set(dessert_users)
    total_unique_users = food_users | dessert_users
    both_users = food_users & dessert_users

    if not total_unique_users:
        raise ValueError("The user lists cannot both be empty")
    return len(both_users) / len(total_unique_users), both_users


zomato_users = ["U1", "U2", "U3", "U4", "U5"]
dessert_users = ["U2", "U4", "U6", "U7"]
both_probability, both_users = probability_of_both(
    zomato_users, dessert_users
)

print(f"Users who ordered both: {sorted(both_users)}")
print(f"Probability of both food and dessert: {both_probability:.2f}")
```

```python
# 5. Conditional probability for Flipkart recommendations

def conditional_probability(event_users, condition_users):
    condition_users = set(condition_users)
    event_users = set(event_users)
    if not condition_users:
        raise ValueError("condition_users cannot be empty")
    return len(event_users & condition_users) / len(condition_users)


phone_buyers = ["U1", "U2", "U3", "U4", "U5"]
headphone_buyers = ["U2", "U3", "U6", "U7"]
probability_headphones_given_phone = conditional_probability(
    headphone_buyers, phone_buyers
)

print(f"P(headphones | phone): {probability_headphones_given_phone:.2f}")
print("Flipkart can recommend headphones to users who bought a phone.")
```
