# Correlation and Scatter Plots

## 1. Instagram Hours and Posts

The following lists represent one person's Instagram activity for seven days. The scatter plot should show a **positive correlation**: days with more hours spent on Instagram generally have more posts.


```python
import matplotlib.pyplot as plt

instagram_hours = [1, 2, 1.5, 3, 2.5, 4, 3.5]
instagram_posts = [2, 4, 3, 7, 5, 9, 8]

plt.scatter(instagram_hours, instagram_posts, color="purple")
plt.xlabel("Instagram hours per day")
plt.ylabel("Instagram posts per day")
plt.title("Instagram Hours vs Posts")
plt.grid(True, alpha=0.3)
plt.show()
```

## 2. Zomato and Swiggy Orders

Example data for six friends is given below because the prompt did not include specific values:

- Monthly Zomato orders: `[2, 4, 5, 7, 9, 10]`
- Monthly Swiggy orders: `[1, 3, 4, 8, 7, 11]`

The correlation is expected to be positive because friends who order more often from Zomato also tend to order more often from Swiggy.

```python
import numpy as np

zomato_orders = [2, 4, 5, 7, 9, 10]
swiggy_orders = [1, 3, 4, 8, 7, 11]

correlation_matrix = np.corrcoef(zomato_orders, swiggy_orders)
correlation = correlation_matrix[0, 1]

print(f"Correlation coefficient: {correlation:.3f}")
print("The correlation is positive." if correlation > 0 else "The correlation is negative." if correlation < 0 else "The correlation is zero.")
```

## 4. Spotify Time and Test Marks

The following generated dataset shows a strong negative correlation: as time spent on Spotify increases, test marks generally decrease.

| Spotify hours per day | Test marks |
|---:|---:|
| 1.0 | 95 |
| 2.0 | 90 |
| 2.5 | 88 |
| 3.0 | 82 |
| 3.5 | 78 |
| 4.0 | 72 |
| 5.0 | 65 |
| 6.0 | 58 |
| 7.0 | 50 |
| 8.0 | 42 |


```python
import matplotlib.pyplot as plt

spotify_hours = [1.0, 2.0, 2.5, 3.0, 3.5, 4.0, 5.0, 6.0, 7.0, 8.0]
test_marks = [95, 90, 88, 82, 78, 72, 65, 58, 50, 42]

plt.scatter(spotify_hours, test_marks, color="green")
plt.xlabel("Time spent on Spotify (hours per day)")
plt.ylabel("Test marks")
plt.title("Spotify Time vs Test Marks")
plt.grid(True, alpha=0.3)
plt.show()
```
