# Population, Sample, and Sampling

## 1. Population and Sample Examples

1. **Instagram followers:** The population is every follower of a particular Instagram account. A sample could be 100 followers selected randomly to answer questions about the account's content.
2. **Zomato restaurants:** The population is all restaurants listed on Zomato in a chosen city. A sample could be 50 randomly selected restaurants whose delivery ratings are studied.
3. **Flipkart products:** The population is all products listed in a category, such as smartphones. A sample could be 75 selected products used to analyze prices, ratings, or customer reviews.

## 2. Random Sampling Simulation

The code cell below creates 1,000 fake user IDs and uses `random.sample()` to select 50 users without replacement.


```python
import random

user_ids = [f"user{i}" for i in range(1, 1001)]
survey_sample = random.sample(user_ids, 50)

print("Selected user IDs:")
print(survey_sample)
```
