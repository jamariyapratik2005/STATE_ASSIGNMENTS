# Population, Sample, and Sampling

## 1. Population and Sample Examples

1. **Instagram followers:** The population is every follower of a particular Instagram account. A sample could be 100 followers selected randomly to answer questions about the account's content.
2. **Zomato restaurants:** The population is all restaurants listed on Zomato in a chosen city. A sample could be 50 randomly selected restaurants whose delivery ratings are studied.
3. **Flipkart products:** The population is all products listed in a category, such as smartphones. A sample could be 75 selected products used to analyze prices, ratings, or customer reviews.

## 2. Random Sampling Simulation

The code cell below creates 1,000 fake user IDs and uses `random.sample()` to select 50 users without replacement.


```python
import random

user_ids = [f"user{i}" for i in range(1, 1001)]
survey_sample = random.sample(user_ids, 50)

print("Selected user IDs:")
print(survey_sample)
```

## 3. Why Use Sampling for Swiggy Delivery Times?

Sampling is faster, cheaper, and easier to manage than collecting delivery-time data from every order ever made. A well-chosen sample can provide a reliable estimate while reducing the time and storage needed for analysis.

## 4. Bias in the Spotify Survey

This introduces **selection bias**, specifically **undercoverage bias**. Because the survey is sent only to users who recently streamed Bollywood songs, users with other music preferences, including people who rarely or never stream Bollywood music, are excluded. The results will therefore overrepresent Bollywood listeners and may not describe Spotify's entire user base.

## 5. Sampling BookMyShow Users

- **Stratified sampling:** Divide users into groups such as age ranges, cities, or booking frequency. Randomly select users from each group in proportion to its size so that every important group is represented.
- **Systematic sampling:** Arrange BookMyShow users in a list, choose a random starting position, and then select every 100th user. This gives a spread-out sample across the list, provided the list does not contain a repeating pattern related to movie-booking habits.
