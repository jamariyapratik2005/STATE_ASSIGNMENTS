```python
print("2. Music App Data Classification:")

music_data = ['Pop', 'Rock', 'Jazz', 'Hip-Hop', 'Pop', 'Rock', 'Jazz', 'Pop']

print(f"Data: {music_data}")
print("Classification: Categorical (Nominal) Data")
print("Reason: The data represents music genres, which are labels or categories.")
print("There is no natural order or ranking among Pop, Rock, Jazz, and Hip-Hop.")

print("\n")
```

```python
print("3. Improving Zomato Restaurant Ratings using Statistics:")

ratings = [5, 4, 5, 5, 4, 3, 1]

mean_rating = statistics.mean(ratings)
median_rating = statistics.median(ratings)

print(f"Sample Ratings: {ratings}")
print(f"Mean Rating: {mean_rating:.2f} stars")
print(f"Median Rating: {median_rating} stars")

print("Explanation:")
print("Statistics can help analyze restaurant ratings using measures such as mean and median.")
print("This helps users understand the overall quality of a restaurant more reliably.")
print("The app can also analyze rating patterns to improve restaurant rankings and recommendations.")

print("\n")
```

```python
print("4. Statistical Terms for Spotify Recommendations:")

# Example user listening data
song_bpms = [120, 115, 120, 130, 125]
listening_minutes = [30, 45, 60, 45, 50]

user_mean_bpm = statistics.mean(song_bpms)
user_median_minutes = statistics.median(listening_minutes)
user_mode_bpm = statistics.mode(song_bpms)

print("- Mean (Average):")
print(f"  Example: The user's average song tempo is {user_mean_bpm} BPM.")
print("  This can help recommend playlists with a similar tempo.")

print("- Median (Middle Value):")
print(f"  Example: The user's median listening time is {user_median_minutes} minutes.")
print("  This can help recommend playlists suitable for the user's typical listening duration.")

print("- Mode (Most Frequent Value):")
print(f"  Example: The most frequent song tempo in the user's listening data is {user_mode_bpm} BPM.")
print("  This can help recommend songs with a similar tempo.")
```
