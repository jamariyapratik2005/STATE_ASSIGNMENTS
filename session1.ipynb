```python
import statistics

print("--- Assignment: Intro to Statistics and Data Science ---\n")

print("1. Daily Actions & Data Types:")

daily_actions = [
    {"app": "Instagram", "action": "Like a post", "data_type": "Categorical (Yes/No)"},
    {"app": "Zomato", "action": "Paying bill amount", "data_type": "Numerical (Float)"},
    {"app": "Flipkart", "action": "Choosing payment method", "data_type": "Categorical (Text)"},
    {"app": "Instagram", "action": "Time spent scrolling", "data_type": "Numerical (Integer)"},
    {"app": "Zomato", "action": "Filtering by cuisine", "data_type": "Categorical (Text)"}
]

for item in daily_actions:
    print(f"- {item['app']}: {item['action']} -> {item['data_type']}")
print("\n")
```

```python
print("2. Music App Data Classification:")
music_data = ['Pop', 'Rock', 'Jazz', 'Hip-Hop', 'Pop', 'Rock', 'Jazz', 'Pop']

genre_counts = {genre: music_data.count(genre) for genre in set(music_data)}

print(f"Data: {music_data}")
print("Classification: Categorical (Nominal) Data")
print("Reason: These are text labels without any mathematical order. We can count their frequencies, but cannot perform math operations like mean.")
print(f"Frequencies: {genre_counts}\n")
```

```python
print("3. Improving Zomato Ratings using Statistics:")
ratings = [5, 4, 5, 5, 4, 1, 1]

simple_mean = statistics.mean(ratings)
most_common_rating = statistics.mode(ratings)

print(f"Sample Ratings: {ratings}")
print(f"Simple Average (Mean): {simple_mean:.2f} stars (Skewed by fake 1-star reviews)")
print(f"Most Common Rating (Mode): {most_common_rating} stars")
print("Explanation: Using the 'Mode' or ignoring extreme low scores helps show the real quality of the restaurant, ignoring fake outlier reviews.\n")
```

```python
print("4. Statistical Terms for Spotify Recommendations:")

# Dummy data for a user
song_bpms = [120, 115, 120, 130, 125] # Beats per minute
release_years = [2020, 2021, 2022, 2021, 1980] # One old song outlier

user_mode = statistics.mode(music_data) # Using music data from Task 2
user_mean_bpm = statistics.mean(song_bpms)
user_median_year = statistics.median(release_years)

print("- Mode (Most Frequent):")
print(f"  Example: The user's most played genre is '{user_mode}'. Recommend a {user_mode} playlist.")

print("- Mean (Average):")
print(f"  Example: Average tempo of their songs is {user_mean_bpm} BPM. Recommend a medium-paced workout playlist.")

print("- Median (Middle Value):")
print(f"  Example: Median release year of liked songs is {user_median_year} (ignoring the 1980 outlier). Recommend recent hits.")
```
