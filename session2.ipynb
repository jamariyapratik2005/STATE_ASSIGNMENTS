```python
import statistics

# List of daily Spotify song play counts
plays = [120, 135, 150, 200, 120, 90, 200]

# Calculate the mean (average)
mean_plays = statistics.mean(plays)

print("Daily Play Counts:", plays)
print(f"Mean (Average) Number of Plays: {mean_plays:.2f}")
```

```python
# Zomato delivery times list
delivery_times = [30, 25, 40, 35, 30, 45, 30]

# Calculate median
median_time = statistics.median(delivery_times)

print(f"All Delivery Times: {delivery_times}")
print(f"Median Delivery Time: {median_time} minutes")
```

```python

# Function define kiya jo ratings lega
def most_common_rating(ratings):
    return statistics.mode(ratings)

# Flipkart ratings array
flipkart_ratings = [5, 4, 4, 3, 5, 4, 2, 4]

# Function ko call kiya aur result store kiya
mode_rating = most_common_rating(flipkart_ratings)

print(f"Product Ratings: {flipkart_ratings}")
print(f"Most Frequent Rating (Mode): {mode_rating} Stars")
```

```python
import statistics

channel_A = [105, 120, 105, 110, 115]  
channel_B = [200, 210, 205, 190, 205]  
channel_C = [105, 120, 105, 110, 5000] # Channel with outlier

channels = {
    "Channel A": channel_A, 
    "Channel B": channel_B, 
    "Channel C": channel_C
}

for name, views in channels.items():
    c_mean = statistics.mean(views)
    c_median = statistics.median(views)
    c_mode = statistics.mode(views)
    
    print(f"--- {name} ---")
    print(f"Views: {views}")
    print(f"Mean: {c_mean:.2f} | Median: {c_median} | Mode: {c_mode}\n")

print("👉 CONCLUSION:")
print("Channel C is most affected by the outlier 5000.")
print("The outlier increases the mean to 1088.00, while the median remains 110.")
print("This shows that the mean is highly sensitive to extreme values, whereas the median is more resistant to outliers.")
```

```python
import pandas as pd

url = "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/tips.csv"
df = pd.read_csv(url)

print("✅ Dataset successfully loaded! Pehli 5 rows dekhiye:\n")
print(df.head(), "\n")

# Hum 'total_bill' wale column (numeric) ka stats nikalenge
bill_mean = df['total_bill'].mean()
bill_median = df['total_bill'].median()
# Mode multiple values return kar sakta hai, isliye [0] likha taaki pehli value mile
bill_mode = df['total_bill'].mode()[0] 

print(f"========== Statistics for 'Total Bill' ==========")
print(f"Mean (Average Bill) : ${bill_mean:.2f}")
print(f"Median (Middle Bill): ${bill_median:.2f}")
print(f"Mode (Common Bill)  : ${bill_mode:.2f}")
```
