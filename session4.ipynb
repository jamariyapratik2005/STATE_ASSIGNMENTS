```python
# Given the daily order counts for a Swiggy delivery partner over 10 days: [12, 15, 17, 14, 13, 16, 200, 18, 14, 15], plot a simple graph (hand-drawn or using any tool) and describe whether the data is left skewed, right skewed, or symmetrical. Explain your reasoning based on the shape.

import matplotlib.pyplot as plt

orders = [12, 15, 17, 14, 13, 16, 200, 18, 14, 15]

plt.hist(orders, bins=15, color='skyblue', edgecolor='black')
plt.title('Swiggy Daily Order Counts')
plt.xlabel('Number of Orders')

plt.ylabel('Frequency')
plt.show()

"The data is right-skewed. Most values are clustered between 12 and 18, but one very large value, 200, creates a long tail on the right side of the graph. This makes the distribution uneven and not symmetrical."
```

```python
# Take the following list of Instagram influencer follower counts (in thousands): [5, 7, 8, 8, 9, 10, 12, 15, 95]. Calculate the mean and median, then state which is higher and what that tells you about the skewness of the data.If the mean is pulled away from the median, the data is skewed in that direction.

import statistics

followers = [5, 7, 8, 8, 9, 10, 12, 15, 95]

mean_val = statistics.mean(followers)
median_val = statistics.median(followers)

print(f"Mean: {mean_val:.2f}")
print(f"Median: {median_val}")

print("The mean is higher than the median, so the data is right-skewed.")
print("This indicates that a very large value, 95, has pulled the mean upward.")
```

```python
# Find a real-world example of right-skewed data from any app you use (like Zomato order amounts, Paytm wallet balances, or YouTube video views). Briefly describe the dataset and explain why you think it is right skewed.

import matplotlib.pyplot as plt

# Example Zomato order amounts
order_amounts = [200, 250, 300, 350, 400, 450, 500, 600, 700, 2500]

# Plot histogram
plt.hist(order_amounts, bins=5)
plt.xlabel("Order Amount (₹)")
plt.ylabel("Number of Orders")
plt.title("Zomato Order Amounts - Right-Skewed Data")
plt.show()

# Explanation: Most orders are between ₹200 and ₹700, while a few high-value orders such as ₹2,500 create a long tail toward the right. Therefore, the data is right-skewed (positively skewed).
```

```python
# Given this dataset of Flipkart product ratings: [3, 3, 4, 4, 4, 5, 5, 5, 5, 5], draw a rough graph and state if the distribution is symmetrical, left skewed, or right skewed. Justify your answer in one line 

import matplotlib.pyplot as plt

ratings = [3, 3, 4, 4, 4, 5, 5, 5, 5, 5]

# Plotting the frequencies
plt.hist(ratings, bins=[2.5, 3.5, 4.5, 5.5], color='lightgreen', edgecolor='black')
plt.xticks([3, 4, 5])
plt.title('Flipkart Product Ratings')
plt.xlabel('Star Rating')
plt.ylabel('Frequency')
plt.show()

print("The distribution is left-skewed because most ratings are high (4 and 5), while fewer ratings are low (3).")
```
