```python
#  1. Calculating Variance by Hand (Daily Steps)

# **Given Data:** `[4500, 7000, 5000, 8000, 4000, 9000, 6000]`  
# **Total values (n):** 7

# **Step 1: Find the Mean (x̄)**
# (4500 + 7000 + 5000 + 8000 + 4000 + 9000 + 6000) / 7 = 43500 / 7 ≈ 6214.29

# **Step 2: Subtract the Mean from each value (x - x̄)**
# * 4500 - 6214.29 = -1714.29
# * 7000 - 6214.29 = 785.71
# * 5000 - 6214.29 = -1214.29
# * 8000 - 6214.29 = 1785.71
# * 4000 - 6214.29 = -2214.29
# * 9000 - 6214.29 = 2785.71
# * 6000 - 6214.29 = -214.29

# **Step 3: Square each result (x - x̄)²**
# * (-1714.29)² ≈ 2938790
# * (785.71)² ≈ 617340
# * (-1214.29)² ≈ 1474500
# * (1785.71)² ≈ 3188760
# * (-2214.29)² ≈ 4903080
# * (2785.71)² ≈ 7760180
# * (-214.29)² ≈ 45920

# **Step 4: Sum all the squared values**
# Sum = 2938790 + 617340 + 1474500 + 3188760 + 4903080 + 7760180 + 45920 = 20,928,570

# **Step 5: Divide by n - 1 (Sample Variance Formula)**
# Variance = 20928570 / 6 = 3,488,095

# **Final Variance = 3,488,095 steps²**
```

```python
import statistics

# Function to calculate standard deviation
def calculate_standard_deviation(scores):
    std_dev = statistics.stdev(scores)
    return round(std_dev, 2)

# Testing the function
exam_scores = [85, 90, 78, 92, 88]
result = calculate_standard_deviation(exam_scores)

print(f"Exam Scores: {exam_scores}")
print(f"Standard Deviation: {result}")
```

```python
import statistics

friend_a = [200, 200, 200, 200, 200]
friend_b = [100, 300, 150, 400, 50]

std_dev_a = statistics.stdev(friend_a)
std_dev_b = statistics.stdev(friend_b)

print(f"Friend A Standard Deviation: {std_dev_a:.2f}")
print(f"Friend B Standard Deviation: {std_dev_b:.2f}\n")

print("👉 EXPLANATION:")
print("Friend A has more consistent spending because their standard deviation is exactly 0.00.")
print("This means Friend A's spending never deviates from their average of 200.")
print("Friend B's high standard deviation (145.77) shows highly unpredictable daily spending.")
```

```python
import statistics

salaries = [28000, 29000, 27000, 60000, 26500, 27500]

variance_salary = statistics.variance(salaries)
std_dev_salary = statistics.stdev(salaries)

print(f"Salaries: {salaries}")
print(f"Variance: {variance_salary:.2f}")
print(f"Standard Deviation: {std_dev_salary:.2f}\n")

print("👉 INTERPRETATION:")
print("The high standard deviation (approx 13,255) indicates a massive spread in the data due to a single outlier (60,000).")
print("This extreme spread means the average salary is misleading, as the majority of the team actually earns between 26,500 and 29,000.")
```
