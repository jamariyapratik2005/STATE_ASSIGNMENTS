# Bayes Theorem in Real-World Applications

This notebook solves four Bayes theorem problems using Python.

```python
# 1. Calculate a posterior probability using Bayes' theorem

def bayes_posterior(prior, likelihood, evidence):
    """Return P(H|E) = P(E|H) * P(H) / P(E)."""
    if evidence <= 0:
        raise ValueError("evidence must be greater than zero")
    return (likelihood * prior) / evidence


prior = 0.30
likelihood = 0.80
evidence = 0.50
posterior = bayes_posterior(prior, likelihood, evidence)

print(f"Posterior probability: {posterior:.2f}")
```

```python
# 2. Swiggy: probability that a discounted restaurant is premium

def bayes_posterior(prior, likelihood, evidence):
    return (likelihood * prior) / evidence


# Prior: P(Premium) = 10% of all restaurants
prior_premium = 0.10
# Likelihood: P(Discount | Premium) = 80%
likelihood_discount_given_premium = 0.80
# P(Discount) = P(Discount | Premium)P(Premium)
#                 + P(Discount | Regular)P(Regular)
prior_regular = 1 - prior_premium
likelihood_discount_given_regular = 0.20
evidence_discount = (
    likelihood_discount_given_premium * prior_premium
    + likelihood_discount_given_regular * prior_regular
)

premium_given_discount = bayes_posterior(
    prior_premium,
    likelihood_discount_given_premium,
    evidence_discount
)
print(f"P(Premium | Discount): {premium_given_discount:.2f}")
```

```python
# 3. Spam detection using the word "free"

def bayes_posterior(prior, likelihood, evidence):
    return (likelihood * prior) / evidence


email_subjects = [
    ("Free prize waiting for you", "spam"),
    ("Team meeting at 10 AM", "not_spam"),
    ("Get free cash today", "spam"),
    ("Your order has shipped", "not_spam"),
    ("Free vacation offer", "spam"),
    ("Project report attached", "not_spam"),
    ("Congratulations, free gift", "spam"),
    ("Family dinner this Sunday", "not_spam"),
    ("Claim your free reward", "spam"),
    ("Your bank statement is ready", "not_spam"),
    ("Free coupon inside", "spam"),
    ("Free meeting notes", "not_spam"),
    ("Limited-time lottery prize", "spam"),
    ("Weekend plans", "not_spam"),
    ("Urgent account verification", "spam"),
    ("Free shipping on your order", "not_spam"),
    ("You won a free bonus", "spam"),
    ("Invoice for your purchase", "not_spam"),
    ("Exclusive free offer", "spam"),
    ("New course schedule", "not_spam")
]

spam_subjects = [subject for subject, label in email_subjects if label == "spam"]
not_spam_subjects = [
    subject for subject, label in email_subjects if label == "not_spam"
]
spam_prior = len(spam_subjects) / len(email_subjects)
free_given_spam = sum("free" in subject.lower() for subject in spam_subjects) / len(spam_subjects)
free_given_not_spam = sum(
    "free" in subject.lower() for subject in not_spam_subjects
) / len(not_spam_subjects)
evidence_free = (
    free_given_spam * spam_prior
    + free_given_not_spam * (1 - spam_prior)
)

spam_given_free = bayes_posterior(
    spam_prior, free_given_spam, evidence_free
)
print(f"P(Spam | 'free'): {spam_given_free:.2f}")
```

## 4. App Example and Prompt Used

**Prompt used with ChatGPT/Copilot:**

> Create a Bayes theorem example for Flipkart. Suppose 30% of users buy a smartphone, 60% of smartphone buyers also buy a phone cover, and 10% of users who do not buy a smartphone buy a phone cover. Explain how to calculate the probability that a user bought a smartphone given that they bought a phone cover.

```python
# 4. Flipkart recommendation example solved with Bayes' theorem

def bayes_posterior(prior, likelihood, evidence):
    return (likelihood * prior) / evidence


# Prior: P(Smartphone) = 30%
prior_smartphone = 0.30
# Likelihood: P(Cover | Smartphone) = 60%
likelihood_cover_given_smartphone = 0.60
# P(Cover | Not Smartphone) = 10%
likelihood_cover_given_not_smartphone = 0.10
prior_not_smartphone = 1 - prior_smartphone

evidence_cover = (
    likelihood_cover_given_smartphone * prior_smartphone
    + likelihood_cover_given_not_smartphone * prior_not_smartphone
)
smartphone_given_cover = bayes_posterior(
    prior_smartphone,
    likelihood_cover_given_smartphone,
    evidence_cover
)

print(f"P(Smartphone | Phone cover): {smartphone_given_cover:.2f}")
print("Flipkart can recommend smartphone accessories using this result.")
```
