# Normal Distribution in Real-World Data

This notebook demonstrates simulation, summary statistics, and applications of the normal distribution.

```python
# 1. Generate and plot 100 normally distributed values
import numpy as np
import matplotlib.pyplot as plt


np.random.seed(42)
mean = 50
standard_deviation = 10
random_numbers = np.random.normal(mean, standard_deviation, 100)

plt.hist(random_numbers, bins=12, density=True, edgecolor="black", alpha=0.7,
         label="Generated values")
x_values = np.linspace(20, 80, 300)
normal_curve = (
    np.exp(-0.5 * ((x_values - mean) / standard_deviation) ** 2)
    / (standard_deviation * np.sqrt(2 * np.pi))
)
plt.plot(x_values, normal_curve, color="red", linewidth=2,
         label="Normal bell curve")
plt.xlabel("Value")
plt.ylabel("Density")
plt.title("Normal Distribution: Mean 50, Standard Deviation 10")
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```

```python
# 2. Calculate the mean, median, and mode of 20 daily step counts
import numpy as np
from scipy import stats


daily_steps = np.array([
    6500, 7200, 8000, 7600, 9000,
    6800, 7500, 8200, 7900, 7000,
    7600, 8400, 7300, 7800, 8100,
    6900, 7700, 7600, 8500, 7400
])

mean_steps = np.mean(daily_steps)
median_steps = np.median(daily_steps)
mode_result = stats.mode(daily_steps, keepdims=False)
mode_steps = mode_result.mode

print(f"Mean: {mean_steps:.2f} steps")
print(f"Median: {median_steps:.2f} steps")
print(f"Mode: {mode_steps:.2f} steps")
print("Mean, median, and mode are approximately equal." if
      abs(mean_steps - median_steps) <= 300 and
      abs(mean_steps - mode_steps) <= 300 else
      "The three values are not approximately equal.")
```

```python
# 3. Simulate 200 online-game scores and count scores within one standard deviation
import numpy as np


np.random.seed(42)
mean_score = 70
standard_deviation_score = 15
scores = np.random.normal(mean_score, standard_deviation_score, 200)

lower_limit = mean_score - standard_deviation_score
upper_limit = mean_score + standard_deviation_score
within_one_standard_deviation = np.sum(
    (scores >= lower_limit) & (scores <= upper_limit)
)
percentage_within_one_standard_deviation = (
    within_one_standard_deviation / len(scores) * 100
)

print(f"Scores between {lower_limit} and {upper_limit}: "
      f"{within_one_standard_deviation} out of {len(scores)}")
print(f"Percentage within one standard deviation: "
      f"{percentage_within_one_standard_deviation:.1f}%")
print("The 68-95-99 rule predicts approximately 68% within one standard deviation.")
```

For a large number of similar orders, most deliveries may be close to the average time, while very fast or very slow deliveries are less common.
